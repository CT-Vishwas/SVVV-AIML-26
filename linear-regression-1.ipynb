Automobile Emission Prediction using Linear Regression

In this notebook, we will build a linear regression model to predict automobile emissions based on various input features such as engine size, fuel type, and vehicle weight. We will use the Mean Squared Error (MSE) as the loss function to evaluate the performance of our model.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
import os
from pathlib import Path
```

```python
current = Path.cwd()
fname = 'vehicle-co2.csv'
file_path = Path.joinpath(current,"data",fname)
```

```python
if not file_path.exists():
    print("File Does not exist")
```

```python
df = pd.read_csv(file_path)
df.head()
```

```python
df.columns
```

```python
df.describe()
```

```python
df['Fuel Type'].value_counts()
```

```python
X = df[['Engine Size(L)', 'Cylinders' ]]
Y = df['CO2 Emissions(g/km)']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
print(f"X_train Shape: {X_train.shape}, X_test Shape: {X_test.shape}")
```

```python
model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
scores = mean_squared_error(Y_test, y_pred)
print(f"MSE: {scores}")
```

```python
prediced_value = model.predict([[2, 4]])
print(f"Co2 Emissions for a 2 Ltr, 4 Cylinder Engine is: {prediced_value[0]:.2f} g/km")
```
