## Analyze the effect of studytime and the Grade

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='studytime', y='Total_Score')
plt.title('Relationship between Study Time and Total Grade')
plt.xlabel('Study Time')
plt.ylabel('Total Score')
```

```python
df['studytime'].value_counts()
```

```python
plt.figure(figsize=(8,6))
sns.regplot(data=df, x='studytime', y='Total_Score')
plt.title('Relationship between Study Time and Total Grade')
plt.xlabel('Study Time')
plt.ylabel('Total Score')
```

Increase in study time is expected to have a positive effect on the grades of students. The increase is significant but not huge enough to make a huge difference in the grades. 

Not all graphs bring out the relationship clearly. It is important to choose the appropriate type of visualization based on the nature of the data and the relationship being analyzed.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x='studytime', y='Total_Score')
plt.title('Relationship between Study Time and Total Grade')
plt.xlabel('Study Time')
plt.ylabel('Total Score')
```

```python
pvt_fjob = df.pivot_table(index='Fjob', values='Total_Score', aggfunc='mean')
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(pvt_fjob, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Relationship between Study Time and Total Grade')
plt.xlabel('Study Time')
plt.ylabel('Fathers Job')
```

The highest grades are for students whose father is a teacher. Those with fathers at home have lowest average grades.
