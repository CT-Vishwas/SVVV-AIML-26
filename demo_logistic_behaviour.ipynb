```python
x = [i/2 for i in range(-8,9)]
```

```python
x
```

```python
import math
def sigmoid(x):
    return 1 / (1+math.exp(-x))
```

```python
previous_y = None
previous_x = None
for item in x:
    y = sigmoid(item)

    if previous_y is None:
        rate_of_change = None
    else:
        rate_of_change = (y - previous_y) / (item - previous_x)

    print(f"x:{item:4.1f}, sigmoid = {sigmoid(item):.4f}, rate_of_change = {rate_of_change if rate_of_change is None else f'{rate_of_change:.4f}'} ")
    previous_x = item
    previous_y = y
```
