## Preprocessing or cleaning
1. There are any missing values           
    a. Drop the row         
    b. Fill with default value
2. Data Type Formats:          
    Integers, Date Formats

```python
d3 = {
    "names": ["vishwas","alice","john","jack","Radha"],
    "city": ["pune","bangalore", "hyderabad","chennai","Kochi"],
    "age": [23,24,25,None,"32"],
    "hiredate": ["23/08/2021","12/09/2020",None,"1/9/2021", None]
}

df3 = pd.DataFrame(d3)
df3
```

```python
df3.isna().sum()
```

Drop the rows

```python
df4 = df3.copy()
df4.dropna()
```

When we drop the rows, we will loose the data. If the data is less loosing the data cannot be afforded.

### Filling with default values
1. 0 as minimum for numerical values
2. Means/Median for numerical values
3. "","Notavailable","unknown" for string columns

```python
df5 = df3.copy()
df5["age"] = df5["age"].fillna(0)
df5["hiredate"] = df5["hiredate"].fillna("1/1/2021")
df5
```

```python
df5.info()
```

```python
df6 = df5.copy()
df6["age"] = df6["age"].astype(int)
df6["hiredate"] = pd.to_datetime(df6["hiredate"], dayfirst=True)
df6.info()
```
